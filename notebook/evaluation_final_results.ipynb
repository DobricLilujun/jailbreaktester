```python
import pandas as pd 

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score


input_folder_path = "/home/snt/projects_lujun/jail/jailbreaktester/output/benchmark_dataset/benchmark_exp_results/JailGuard_processed"

input_file_path = "benchmark_gemma_merged_RandomInsertPerturbation_25_with_only_pert2detect_formalized_23-0241_processed.jsonl"

df = pd.read_json(f"{input_folder_path}/{input_file_path}", lines=True)


y_true = df['manual_checking_label']
y_pred = df['cls_label'].astype(int)

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

print(f'Accuracy: {accuracy:.4f}')
print(f'Precision: {precision:.4f}')
print(f'Recall: {recall:.4f}')
print(f'F1 Score: {f1:.4f}')
```
