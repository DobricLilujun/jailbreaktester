```python
import pandas as pd
import json

df = pd.read_json(
    "/home/snt/projects_lujun/jail/jailbreaktester/output/benchmark_dataset/benchmark_exp_results/negbleurtForest_processed_corrected/negbleurtForest_processed_evaluated_single_prompts/benchmark_llama_merged_RandomSwapPerturbation_25_with_only_pert2detect_formalized_23-0825_processed_14-1013_processed_evaluated.jsonl",
    lines=True,
)

df["real_responses"] = df["responses"].apply(
    lambda x: (
        json.loads(x)[1]
        if isinstance(json.loads(x), list) and len(json.loads(x)) > 1
        else None
    )
)

short_df = df[
    [
        "real_responses",
        "target_responses_summrized",
        "manual_checking_label",
        "jial_break_label",
    ]
]
```

```python
from sklearn.metrics import confusion_matrix

idx = short_df[
    (short_df["jial_break_label"] == True) & (short_df["manual_checking_label"] == 0)
].index

# 将这些索引对应行的 manual_checking_label 设置为 1
short_df.loc[idx, "manual_checking_label"] = 1
y_true = short_df["manual_checking_label"].to_numpy()
y_pred = short_df["jial_break_label"].to_numpy()

cm = confusion_matrix(y_true, y_pred)
print(cm)

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred)
recall = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")
```
