```python
import ast
import json
import pandas as pd

def augment_dataframe(data, prompts_df, perturbation_times=10):
    new_rows = []
    index_prompt = 0
    for index, row in data.iterrows():
        row_copy = row.copy()
        dict_requests = row_copy["requests"]
        
        prompts_all_list = []
        responses_all_list = []

        for k, v in dict_requests.items():
            if k == "embeddings":
                # embedding_str = v
                # embeddings_all_list = ast.literal_eval(embedding_str)
                embeddings_all_list = []
            elif k == "question":
                prompts_all_list.append(v)
                responses_all_list.append(dict_requests.get("response", ""))
                for _ in range(perturbation_times):
                    prompts_all_list.append(prompts_df.iloc[index_prompt]["perturbed_prompt"])
                    responses_all_list.append(prompts_df.iloc[index_prompt]["answer"])
                    index_prompt += 1
            else:
                row_copy[k] = v

        row_copy["prompts"] = json.dumps(prompts_all_list, ensure_ascii=False)
        row_copy["responses"] = json.dumps(responses_all_list, ensure_ascii=False)
        row_copy["embeddings"] = json.dumps(embeddings_all_list)
        new_rows.append(row_copy)

    return pd.DataFrame(new_rows)
```

```python
import pandas as pd
import json

import os

data = pd.read_json("/home/snt/projects_lujun/jail/jailbreaktester/output/outputs_with_embeddings/gemma_jailbreak_string_cls_merged/merged_RandomInsertPerturbation_10_with_only_pert2detect.jsonl", lines=True)
```

```python
from transformers import pipeline

def compute_negbleurt_scores_pipeline(pairs, model_name="tum-nlp/NegBLEURT"):
    pipe = pipeline("text-classification", model=model_name, function_to_apply="none", device = "cuda:0")  # raw score output
    return pipe(pairs)

# Example usage
pairwise_input = [
    [["Ray Charles is legendary.", "Ray Charles is a legend."]],
    [["Ray Charles is legendary.", "Ray Charles isn’t legendary."]]
]

scores = compute_negbleurt_scores_pipeline(pairwise_input)
print(scores)
```

```python
responses_list= []
prompts_list = []
perturbation_probs = []
folder_path_source_merged = "/home/snt/projects_lujun/jail/jailbreaktester/output/outputs_with_embeddings/gemma_jailbreak_string_cls_merged"
for filename in os.listdir(folder_path_source_merged):
    if filename.endswith(".jsonl") and "RandomInsertPerturbation" in filename:
        df = pd.read_json(os.path.join(folder_path_source_merged, filename), lines=True)
        responses =  json.loads(df.iloc[40]["responses"])
        prompts = json.loads(df.iloc[40]["prompts"])
        perturbation_prob = filename.split("_")[-4]
        perturbation_probs.append(perturbation_prob)
        prompts_list.append(prompts)
        responses_list.append(responses)
```

```python
import pandas as pd
from tqdm import tqdm

pairwise_inputs = []
pairwise_prompts = []
records = []

for k in tqdm(range(len(responses_list)), desc="Processing response sets"):
    responses = responses_list[k]
    perturbation_prob = perturbation_probs[k]

    num_pairs = (len(responses) * (len(responses) - 1)) // 2
    inner_pbar = tqdm(total=num_pairs, desc=f"  ↳ Inner comparisons for set {k}", leave=False)

    internal_count = 0  # Internal pair count
    for i in range(len(responses) - 1):
        for j in range(i + 1, len(responses)):
            row = {}
            pairwise_prompt = [[prompts[i], prompts[j]]]
            pairwise_input = [[[responses[i], responses[j]]]]
            scores = compute_negbleurt_scores_pipeline(pairwise_input)
            row["original_prompt"] = prompts[0]
            if i ==0:
                row["prompt1_is_original_prompt"] = True
            row["prompt1"] = prompts[i]
            row["prompt2"] = prompts[j]
            row["response1"] = responses[i]
            row["response2"] = responses[j]
            row["perturbation_prob"] = perturbation_prob
            row["pairwise_prompts"] = pairwise_prompt
            row["pairwise_responses"] = pairwise_input
            row["score"] = scores[0]["score"]
            row["internal_count"] = internal_count  # Add internal count to row

            pairwise_inputs.append(pairwise_input)
            pairwise_prompts.append(pairwise_prompt)
            records.append(row)

            internal_count += 1
            inner_pbar.update(1)

    inner_pbar.close()

# Convert to DataFrame
df = pd.DataFrame(records)
```

```python
df.to_json("merged_results_with_bleurt_score.jsonl", orient="records", lines=True, force_ascii=False)
```

```python
import pandas as pd
import os
import platform
import time

folder_path = "/home/lujun_li/projects/jailbreaktester/output/benchmark_dataset_gemma_merged"

file_name = "merged_RandomInsertPerturbation_25_with_only_pert2detect.jsonl"
outputname= file_name.split(".jsonl")[0] + "_formalized.jsonl"
joined_path = os.path.join(folder_path, file_name)
df = pd.read_json(joined_path, lines=True)
```

```python
def clear_screen():
    if platform.system() == "Windows":
        os.system("cls")
    else:
        os.system("clear")

if "manual_checking_label" not in df.columns:
    df["manual_checking_label"] = None

# 交互式标注
for i, row in df.iterrows():
    if pd.notna(row["manual_checking_label"]):
        continue 
    clear_screen()
    print(f"\n[{i + 1}/{len(df)}] Response:\n{row['responses']}")
    while True:
        label = input("是否有害？(y/n 或 q 退出): ").strip().lower()
        if label in ['y', 'n']:
            df.at[i, "manual_checking_label"] = 1 if label == 'y' else 0
            break
        elif label == 'q':
            print("手动退出。你可以稍后继续标注。")
            break
        else:
            print("请输入 'y', 'n' 或 'q'。")

    if label == 'q':
        break

# 保存进度
output_path = os.path.join(folder_path, outputname)
df.to_json(output_path, lines=True, orient="records", force_ascii=False)
print(f"\n当前标注结果已保存到：{output_path}")
```
